```python
def kub_hesabla(n):
    return n ** 3
print(kub_hesabla(3))
```

```python
def birlesdir(soz1, soz2):
    return soz1 + " " + soz2
print(birlesdir("Salam", "Dünya"))
```

```python
def cap_et(liste):
    for element in liste:
        print(element)
cap_et([1, 2, 3])
```

```python
def toplam(*args):
    return sum(args)
print(toplam(1, 2, 3, 4))
```

```python
def ortalama(*args):
    if len(args) == 0:
        return "Rəqəm yoxdur"
    return sum(args) / len(args)
print(ortalama(10, 20, 30))
```

```python
def adlar_rəqəmlər(**kwargs):
    for ad, reqem in kwargs.items():
        print(f"ad: {ad}, rəqəm: {reqem}")

adlar_rəqəmlər(bir=1, iki=2, uc=3)
```

```python
def tip_yoxla(dəyər):
    if isinstance(dəyər, str):
        print("mətn")
    elif isinstance(dəyər, (int, float)):
        print("rəqəm")
    else:
        print("başqa")
tip_yoxla("test")
```
